```python
import numpy as np
import random
import timeit

%timeit [random.random() for _ in range(100_000)]
%timeit np.random.random(100_000)

N = 100 #number of repetitions
#List comprehension
lc_time = timeit.timeit('[random.random() for _ in range(100_000)]', setup='import random',number=N)

np_time = timeit.timeit('np.random.random(100_000)',setup='import numpy as np',number=N )

lc_avg = lc_time / N * 1000
np_avg = np_time / N * 1000


print(f'List comprehension : {lc_avg:.3f} ms per loop')
print(f'NumPy              : {np_avg:.3f} ms per loop')
print(f'Speedup            : {lc_avg / np_avg:.1f}x faster with NumPy')
```
